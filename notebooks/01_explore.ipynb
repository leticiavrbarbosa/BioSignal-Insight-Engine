```python
import pandas as pd
import matplotlib.pyplot as plt
from sqlalchemy import create_engine
from dotenv import load_dotenv, find_dotenv
import os
```

```python
print(os.getenv("DATABASE_URL"))
```

```python
print(find_dotenv())
```

```python
load_dotenv(override=True)
engine = create_engine(os.getenv("DATABASE_URL"))
```

```python
# query from the DB
df = pd.read_sql("SELECT * FROM physiological_signals LIMIT 1000", engine)
```

```python
# plot the ecg 
plt.figure(figsize=(12,3))
print(df["ecg"].values)
plt.plot(df["ecg"].values)
plt.title("Raw ECG Signal - Sample")
plt.xlabel('Sample')
plt.ylabel("Amplitude")
plt.show()
```
